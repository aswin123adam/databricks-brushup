## Creating an External Location

ex code: 
CREATE EXTERNAL LOCATION IF NOT EXISTS location_name
    URL url_str
    WITH (STORAGE CREDENTIAL credential_name)
    [ COMMENT comment ]

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricks_storage_demo
    URL 'abfss://demo@dbprepdatalake.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL databricks_prep_rg_ext_sc)
    COMMENT 'External Location Connection secured for Databricks'
```

# Creating a Container in Azure ADLS


```python
%fs ls 'abfss://demo@dbprepdatalake.dfs.core.windows.net/demo_dir/'
```

```python
# Read JSON data into DataFrame
df = spark.read.json('abfss://demo@dbprepdatalake.dfs.core.windows.net/demo_dir/orders_2024_10.json')

# Preview the data
display(df.limit(5))

# Save as a managed Delta table in Databricks
df.write.format("delta").mode("overwrite").saveAsTable("orders_2024_10")

print("Delta table 'orders_2024_10' created successfully!")
```
