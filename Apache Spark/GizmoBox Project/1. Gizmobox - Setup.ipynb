#### CODE TO CREATE STORAGE CONNECTION WITH CREDENTIALS


```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS gizmobox
    URL 'abfss://gizmobox@dbprepdatalake.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL databricks_prep_rg_ext_sc)
    COMMENT 'GizmoBox Linked and Setup Completed';
```

# Accessing the new container

```sql
%fs ls 'abfss://gizmobox@dbprepdatalake.dfs.core.windows.net/'
```

Creating the Catalog for Gizmobox

```sql
SHOW CATALOGS;

CREATE CATALOG IF NOT EXISTS gizmobox
    MANAGED LOCATION 'abfss://gizmobox@dbprepdatalake.dfs.core.windows.net/'
    COMMENT 'GizmoBox Catalog';
```

#### Create Schemas
1. Landing
2. Bronze
3. Silver
4. Gold

```sql
SELECT current_catalog();
```

```sql
use catalog gizmobox;

show schemas;

create schema if not exists landing
    MANAGED LOCATION 'abfss://gizmobox@dbprepdatalake.dfs.core.windows.net/landing/';
create schema if not exists bronze
    MANAGED LOCATION 'abfss://gizmobox@dbprepdatalake.dfs.core.windows.net/bronze/';
create schema if not exists silver
    MANAGED LOCATION 'abfss://gizmobox@dbprepdatalake.dfs.core.windows.net/silver/';
create schema if not exists gold
    MANAGED LOCATION 'abfss://gizmobox@dbprepdatalake.dfs.core.windows.net/gold/';
```

Create Volumes

```sql
USE CATALOG gizmobox;
USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
    LOCATION 'abfss://gizmobox@dbprepdatalake.dfs.core.windows.net/landing/operational_data';

-- CREATE VOLUME IF NOT EXISTS external_data
--     LOCATION 'abfss://gizmobox@dbprepdatalake.dfs.core.windows.net/landing/external/data';
```
