Higher Order Functions  (Commonly used!)
- Transform
- Filter
- Exists
- Aggregate

```sql
select * from gizmobox.silver.orders;
```

```sql
CREATE OR REPLACE TEMP VIEW ORDER_EX
AS
SELECT * from VALUES
(1, array('smartphone', 'laptop' , 'monitor')),
(2, array('tablet','smartwatch','headphones')),
(3, array('keyboard','mouse'))
as orders_ex(order_id, items)
```

```sql
select * from order_ex
```

```sql
-- select order_id, explode(items) as item from order_ex
select order_id, 
    transform( items, x -> upper(x))
from order_ex
```

```sql
select order_id,
        filter(items, x -> x LIKE '%smart%')
from order_ex
```
