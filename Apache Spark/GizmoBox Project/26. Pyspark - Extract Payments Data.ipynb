```python
paymentSchema = 'payment_id integer, order_id integer, payment_timestamp timestamp, payment_status integer, payment_method string'
```

```python
df = (
    spark.read.format("csv")
    .option('delimiter',',')
    .schema(paymentSchema)
    .load('abfss://gizmobox@dbprepdatalake.dfs.core.windows.net/landing/external_data/payments')
)
display(df)
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType,TimestampType
py_payments_schema = StructType([
    StructField('payment_id', IntegerType()),
    StructField('order_id', IntegerType()),
    StructField('payment_timestamp', TimestampType()),
    StructField('payment_status', IntegerType()),
    StructField('payment_method', StringType())
])
```

```python
df = (
    spark.read.format("csv")
        .option('delimiter',',')
        .schema(py_payments_schema)
        .load('abfss://gizmobox@dbprepdatalake.dfs.core.windows.net/landing/external_data/payments')
)
display(df)
```
