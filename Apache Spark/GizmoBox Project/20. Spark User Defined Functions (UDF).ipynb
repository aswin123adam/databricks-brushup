####Syntax

CREATE OR REPLACE FUNCTION catalog_name.schema_name.udf_name(param_name.data_type)
RETURNS return_type
RETURN expression

```sql
create or replace function gizmobox.default.get_fullname(firstname string, surname string)
returns string
return concat(initcap(firstname), ' ', initcap(surname));
```

```sql
select gizmobox.default.get_fullname('john', 'doe');
```

```sql
DESC FUNCTION EXTENDED gizmobox.default.get_fullname
```

```sql
select payment_id,
        order_id,
        payment_timestamp,
        date_format(payment_timestamp, 'yyyy-MM-dd') as date,
        date_format(payment_timestamp, 'HH:mm:ss') as time,
        payment_status,
        payment_method
from gizmobox.bronze.payments;
```

```sql
create or replace function gizmobox.default.get_ps(payment_status INTEGER)
returns string
return  case when payment_status = 1 then 'Success'
                when payment_status = 2 then 'Pending'
                 when payment_status = 3 then 'Cancelled'
                 when payment_status = 4 then 'Failed'
        end
```

```sql
select gizmobox.default.get_ps(2);
```

```sql
select payment_id,
        order_id,
        payment_timestamp,
        date_format(payment_timestamp, 'yyyy-MM-dd') as date,
        date_format(payment_timestamp, 'HH:mm:ss') as time,
        payment_status,
        gizmobox.default.get_ps(payment_status) as payment_status_text,
        payment_method
from gizmobox.bronze.payments;
```
