```python
df = spark.table("gizmobox.bronze.py_addresses")
display(df)
```

```python
from pyspark.sql import functions as F
df_pivoted_data = (
    df.groupBy("customer_id")
    .pivot("address_type", ['shipping', 'billing'])
    .agg(
        F.max("address_line_1").alias("address"),
        F.max("city").alias("city"),
        F.max("state").alias("state"),
        F.max("postcode").alias("postcode")
    )
)
display(df_pivoted_data)
```

```python
df_pivoted_data.writeTo("gizmobox.silver.py_addresses").createOrReplace()
```
