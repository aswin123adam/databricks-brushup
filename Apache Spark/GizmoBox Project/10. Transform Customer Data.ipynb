```sql
-- remove record with NULL customer_id

select * from gizmobox.bronze.v_customers
where customer_id is not null;
```

```sql
select distinct(*) from gizmobox.bronze.v_customers
where customer_id is not null
```

```sql
select customer_id,
        max(current_timestamp),
        max(customer_id),
        max(customer_name),
        max(date_of_birth),
        max(email),
        max(member_since),
        max(telephone),
        max(file_path)
from gizmobox.bronze.v_customers
where customer_id is not null
group by customer_id
```

```sql
-- CAST column to correct datatype
WITH CTE_max AS
(
    SELECT customer_id,
        max(current_timestamp) AS max_current_timestamp
    FROM gizmobox.bronze.v_customers
    WHERE customer_id IS NOT NULL
    GROUP BY customer_id
)
SELECT * FROM CTE_max
```
