```sql
select payment_id,
       order_id,
       payment_timestamp,
       payment_status,
       payment_method
from gizmobox.bronze.payments;
```

```sql
-- Extract Date and time from the payment_timestamp

select payment_id,
        order_id,
        payment_timestamp,
        date_format(payment_timestamp, 'yyyy-MM-dd') as date,
        date_format(payment_timestamp, 'HH:mm:ss') as time,
        payment_status,
        payment_method
from gizmobox.bronze.payments;
```

```sql
with cte_mapping as (
    select  payment_id,
            case when payment_status = 1 then 'Success'
                 when payment_status = 2 then 'Pending'
                 when payment_status = 3 then 'Cancelled'
                 when payment_status = 4 then 'Failed'
            else 'Unknown description'
            end as status_description
    from gizmobox.bronze.payments

)

select * from gizmobox.bronze.payments p join cte_mapping cte on p.payment_id = cte.payment_id
order by p.payment_id   
```
