#Extract Data From the Customer JSON Files
- Query Single File
- Query List of Files using wildcard Characters
- Query all the files in a folder

1. Query Single JSON File

```sql
select * from json.`/Volumes/gizmobox/landing/operational_data/customers/customers_2024_10.json`
```

2. Query Muliple JSON with wildcard

```sql
select * from json.`/Volumes/gizmobox/landing/operational_data/customers/customers_2024*.json`
```

3. Query all files in the folder

```sql
select * from json.`/Volumes/gizmobox/landing/operational_data/customers/`
```

SELECT FILE METADATA FOR CODE AUDITABILITY


```sql
select * , _metadata.file_path from json.`/Volumes/gizmobox/landing/operational_data/customers/`
```
